# Sample Size Selection - Proportions

In the previous notebook, we saw various ways to determine the sample size on continuous data. In this notebook, we will look at how to determine required sample size if we are working on proportions.

## Determining Sample Size Based on Standard Error

Previously, we saw that for the sample proportion $\hat{p}$,

Standard error:

$SE = \sqrt{\frac{\hat{p} \times (1 - \hat{p})}{n}}$ 

Error margin:

$E = z^{\star} \times \sqrt{\frac{\hat{p} \times (1 - \hat{p})}{n}}$

where $z^{\star}$ refers to the critical value associated with the desired confidence level.

When we solve this for $n$:

$E = z^{\star} \times \sqrt{\frac{\hat{p} \times (1 - \hat{p})}{n}}$

$\frac{E}{z^{\star}} = \sqrt{\frac{\hat{p} \times (1 - \hat{p})}{n}}$

$\left(\frac{E}{z^{\star}}\right)^2 = \frac{\hat{p} \times (1 - \hat{p})}{n}$

$n = \frac{(z^{\star})^2 \times \hat{p} \times (1 - \hat{p})}{E^2}$

If we have an expected margin of error, we can plug in that value and obtain the required sample size.

**Note:** A question that may arise naturally is how we can use $z$ values instead of $t$, when we do not have a sample to start with. Since there is no prior information while determining the required sample size, traditionally $z$ values are used. Once we get $n$, then we can determine which critical value to use.

**Example**

We hypothesize that selecting chocolate flavored breakfast cereal is slightly more common compared to the standard counterparts. If we model that using $\hat{p} = 0.6$ and expect to get an error margin of 0.2 then,

$\hat{p} = 0.6$

$E = 0.2$

and two-sided critical value for 5% error rate:

$z^* = z_{0.025}$

```python
p = 0.6
variance = p * (1 - p)
alpha = 0.05
z_critical = stats.norm.ppf(1 - alpha / 2)
error_margin = 0.2

n = (z_critical ** 2 * variance) / (error_margin ** 2)
print(f"Required sample size for estimating the proportion with a margin of error of {error_margin}: {int(np.ceil(n))}")
```

## Sample Size Selection Based on Statistical Power

Alternatively, we can aim for at least 80% statistical power to obtain the required sample size. For the following hypothesis:

$H_0: p = p_1$

$H_1: p \neq p_1$

the required sample size can be computed as:

$n = \frac{\left(z^{\star}\sqrt{p_1(1-p_1)} + z_{\beta}\sqrt{p_a(1-p_a)}\right)^2}{(p_a - p_1)^2}$

where $z^{\star}$ is the critical value for the desired confidence level, $z_{\beta}$ is the critical value for the desired statistical power, and $p_a$ is the probability we wish to detect under $H_1$.

**Example**

We hypothesize that chocolate flavored breakfast cereals are a more common choice across cereal consumers compared to the standard counterpart. We want to prove that the real probability $p$ is different from 0.5, but instead it is 0.6. Then, under 80% statistical power expectation and 5% error rate,

$z^* = z_{0.05} \quad \text{(This time we are investigating only one side)}$

$z_{\beta} = 0.84 \quad \text{(For 80\% statistical power)}$

$p_1 = 0.5$

$p_a = 0.6$


```python
alpha = 0.05
z_critical = stats.norm.ppf(1 - alpha)
z_beta = stats.norm.ppf(0.8)
p1 = 0.5
pa = 0.6

n = (z_critical * np.sqrt(p1 * (1-p1)) + z_beta * np.sqrt(pa * (1-pa)))**2 / (pa - p1)**2
print(f"Required sample size for detecting a difference in proportions with 80% power: {np.ceil(n)}")
```
